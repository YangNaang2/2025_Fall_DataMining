Linear Regression with Gradient Descent

데이터에 가장 잘 맞는 직선(y=wx+b)를 찾는다.

핵심 로직
1. 현재 모델(w,b)로 예측값을 구한다.
2. 실제값과의 차이(오차)를 구함
3. 오차를 줄이는 방향(기울기)을 구해 w,b 업데이트

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder\
  .master("local[*]")\
  .getOrCreate()
sc = spark.sparkContext
```

```python
#데이터: (x,y)
data = sc.parallelize([(1,2), (2,4), (3,6), (4,8)])

#파라미터 초기화 (기울기w, 절편b)
w=0.0
b=0.0
learning_rate=0.01
n = data.count()
```

```python
#경사 하강법 시작
for i in range(10):
  #각 노드에 현재 w,b값을 알려줌
  #각 데이터 포인트에서의 기울기를 다 계산(map)
  # 다 더함(reduce)

  #gradient = (예측값-실제값) * x
  gradients = data.map(
      lambda p:(
          (p[0] * w + b -p[1]) * p[0], #w에 대한 미분
          (p[0] * w + b -p[1]) #b에 대한 미분
      )
  ).reduce(lambda a, b: (a[0]+b[0], a[1]+b[1]))

  #update
  w -= learning_rate * (gradients[0]/n)
  b -= learning_rate * (gradients[1]/n)
```

```python
print(f"Result: {w:.2f}x + {b:.2f}")
```
