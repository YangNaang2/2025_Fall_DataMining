

```
 (문서ID, "단어들~~") -> (단어, 문서ID) 형태로 뒤집기
```



````python
from pyspark.sql import SparkSession

spark = SparkSession.builder\
  .master("local[*]")\
  .getOrCreate()
sc = spark.sparkContext
````

````python
docs=[
    (1, "spark is fast"),
    (2, "spark is cool"),
    (3, "python is good")
]
rdd = sc.parallelize(docs)
````

````python
#단어 쪼개기 & 뒤집기
rdd2 = rdd.flatMap(lambda x: [(word, x[0]) for word in set(x[1].split())])
````

````python
#같은 단어끼리 문서 id 모으기
rdd3 = rdd2.groupByKey().mapValues(list).collect()
````

````python
print(rdd3)
````
