```python
wget.download("https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/int-pair-input.txt", "A.txt")
```

```python
rdd1=sc.textFile("A.txt")
```

```python
rdd2 = rdd1.map(
    lambda line: tuple(map(int, line.strip('()').replace(' ', '').split(',')))
)
```

```python
ans1 = rdd2.first()[0]
ans1
```

```python
ans2 = rdd2.map(lambda p:p[0]).max()
ans2
```

```python
ans3 = rdd2.map(lambda p:p[0]).min()
ans3
```

```python
ans4 = rdd2.filter(lambda p:p[0]==1).count()
ans4
```

```python
ans5 = rdd2.filter(lambda p:p[0]==2).count()
ans5
```

```python
ans1+ans2+ans3+ans4+ans5
```
