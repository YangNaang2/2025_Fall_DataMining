```python
from pyspark.sql import SparkSession

spark = SparkSession.builder\
  .master("local[*]")\
  .getOrCreate()
sc = spark.sparkContext
```

```python
import numpy as np

n=2000
B=10
d=2
threshold = 5.
```

```python
from sklearn.datasets import make_blobs
X, _ = make_blobs(n_samples=n, centers=B, n_features=d,
                  random_state=0)
```

```python
rdd = sc.parallelize([ (i, x) for i, x in enumerate(X) ])
```

```python
rdd.collect()
```

```python
#(i_p_i)->(0,1)~(i-1,i), (i,i+1)~(i,n-1)
rdd1 = rdd.flatMap(lambda t:
    [((j, t[0]), t[1]) for j in range(t[0])] +          # (0~i-1, i)
    [((t[0], j), t[1]) for j in range(t[0]+1, n)]       # (i, i+1~n-1)
)
```

```python
rdd1.collect()
```

```python
import numpy as np
rdd2=rdd1.reduceByKey(lambda x,y:np.sqrt(((x-y)*(x-y))).sum())
```

```python
rdd2.collect()
```

```python
rdd2.filter(lambda t: t[1]<threshold).count()
```
